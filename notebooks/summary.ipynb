# Summarizing pupillometric data

`pypillometry` summarizes a dataset at several levels:

- **one dataset**: its contents (`print(d)`, `summary()`), the quality of
  the data (`quality_report()`) and how much of it is masked, and why
  (`layer_stats()`)
- **several datasets**: the same summaries as one table with a row per
  dataset, e.g., to find participants with poor data
- **trials**: one value per event or trial (`stat_per_event()`), together
  with the events and their metadata

Many of these results are dicts that are displayed as a table: per
channel (e.g., `layer_stats()`: channel -> layer -> statistics) or with
one row per trial (a dict of columns). `.to_pandas()` converts them to a
`pandas.DataFrame`.

```python
import pypillometry as pp
from pypillometry import u
import pandas as pd

pp.logging_set_level("WARNING")
```

## One dataset

We detect and interpolate blinks and mask the gaze outside the screen.
Printing the object gives an overview of its contents and the history of
operations:

```python
d = pp.get_example_data("rlmw_002_short")
d = d | pp.pupil.blinks_detect() | pp.pupil.blinks_interpolate() | pp.gaze.mask_offscreen()
print(d)
```

`summary()` returns the same information as a `dict`:

```python
d.summary()
```

### Data quality

The quality report rates missing data, noise, blinks, gaze precision,
the agreement of the eyes and the calibration as "good", "ok", "poor" or
"bad" (see [Data quality](../user_guide/quality.rst)):

```python
report = d.quality_report()
report
```

`report.issues()` lists the measures that are "poor" or "bad" (none here),
and `report.to_dict(flat=True)` gives all values as a flat `dict`, e.g.,
for a table of many datasets (see below).

### Masked data

`layer_stats()` describes each mask layer of each channel: the number of
stretches, their durations and the percentage of the samples they cover.
The entry `masked` counts all unusable samples of a channel (see
[Missing data](../user_guide/missing_data.rst)). It is a dict (channel ->
layer -> statistics) that is displayed as a table:

```python
stats = d.layer_stats(units="sec")
stats
```

```python
stats["left_pupil"]["blink"]["percent"]   # % of the samples in blinks
```

After the interpolation, the pupil channels have no unusable samples left
(`masked`: 0 stretches), because interpolated samples are usable; the gaze is still
masked during blinks and off the screen.

`blink_stats()` is a shortcut for the blink rows. Durations of the
recording are given by `get_duration()` and, if periods are excluded
(e.g., breaks), `excluded_duration()`:

```python
d.blink_stats()["left_pupil"], d.get_duration("min"), d.excluded_duration("min")
```

## Several datasets

In a study, there is one dataset per participant. Here, we use three
example recordings and detect blinks in each of them:

```python
datasets = [pp.get_example_data("rlmw_002_short"),
            pp.EyeData.from_file("../../data/test.pd"),
            pp.EyeData.from_file("../../data/example_shortblinks.pd")]
datasets = [d | pp.pupil.blinks_detect() for d in datasets]
```

A table of their summaries (some of the columns):

```python
tab = pd.DataFrame([d.summary() for d in datasets])
tab[["name", "n", "sampling_rate", "duration_minutes", "nevents", "nblinks"]]
```

The quality reports as a table, one row per dataset, with the overall
label and the missing pupil data of the left eye:

```python
quality = pd.DataFrame([d.quality_report().to_dict(flat=True) for d in datasets])
quality[["name", "overall", "pupil.left.missing_pct", "pupil.left.missing_pct.label",
         "blinks.left.rate", "blinks.left.rate.label"]]
```

Datasets that need a closer look can be selected by their labels, and
`issues()` tells what is wrong with them:

```python
quality[quality["overall"].isin(["poor", "bad"])][["name", "overall"]]
```

```python
datasets[2].quality_report().issues()
```

The layer statistics of all datasets in one table, with a column for the
dataset:

```python
pd.concat([d.layer_stats(variables="pupil").to_pandas().reset_index().assign(name=d.name)
           for d in datasets], ignore_index=True)
```

## Trials

`stat_per_event()` computes one value per event, e.g., the mean pupil
size in the second after each stimulus. Combined with the events and
their metadata (see [Event metadata](event_metadata.ipynb)), this gives a
table with one row per trial, e.g., for a statistical model:

```python
d = datasets[0]
epochs = d.get_epochs("S_", (0, 1000) * u.ms)
means = d.stat_per_event(epochs, variables="pupil")
trials = pp.Table({**epochs.events.to_table(), **means})
trials
```

For segments around events (see the
[ERPD tutorial](erpds.ipynb)), `summary()` gives their shape and the
missing data per channel; the measures of gaze in regions of interest per
trial (`seg.gaze.dwell_time()`, `seg.gaze.first_entry()`) are tables like
the one above (see [Regions of interest](rois.ipynb)):

```python
seg = d.get_segments(d.get_epochs("S_", (-500, 2500) * u.ms), variables="pupil")
seg.summary()
```
